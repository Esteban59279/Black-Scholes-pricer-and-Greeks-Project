```python
# Black Scholes formula and risk-sensitivity measures (Greeks)
# Important infos : -> norm.cdf() means cumulative distribution function, it gives the probability that a normally distributed variable is less than or equal to a given value.
#                   -> norm.pdf() means the probability density function, it means the height between the bell curve and the point belonging to the x-axis for a given point. Higher is the PDF value, higher is the probability.
import numpy as np
from scipy.stats import norm
```

```python
from vollib.black_scholes import black_scholes as bs
from vollib.black_scholes.greeks.analytical import delta, gamma, theta, vega, rho

# 1. Define parameters
r = 0.01 # Risk-free rate
S = 30 # Underlying's price
K = 40 # Strike price
T = 240/365 # Time to maturity (before deadline)
sigma = 0.30 # Volatility

# 2. Black Scholes formula

def blackscholes(r,s,K,T,sigma, type="c"):
  "Calculate call/put prices"
  d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  try:
    if type == "c": # if it's a call
      price = S*norm.cdf(d1,0,1) - K*np.exp(-r*T)*norm.cdf(d2,0,1)
    elif type =="p": # if it's a put
      price = K*np.exp(-r*T)*norm.cdf(-d2,0,1) - S*norm.cdf(-d1,0,1)
    return price, bs(type, S, K, T, r, sigma)
  except:
    print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
print("Option price: ", blackscholes(r, S, K, T, sigma,"c"))  # Option price if it's a call for example
```

```python
# 3. Risk-sensitivity measures (Greeks)
# 3.1 Delta calculus --- sensitivity of option price to changes in the underlying asset price.

def delta_calc(r,s,K,T,sigma, type="c"):
    "Calculate Delta of an option"
    d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
    try:
      if type == "c": # if it's a call
       delta_calc = norm.cdf(d1,0,1)
      elif type =="p": # if it's a put
       delta_calc = -norm.cdf(-d1,0,1)
      return delta_calc, delta(type, S, K, T, r, sigma)
    except:
      print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
```

```python
# 3.2 Gamma calculus --- sensitivity of Delta to changes in the underlying price.

def gamma_calc(r,s,K,T,sigma, type="c"):
  "Calculate Gamma of an option"
  d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  try:
    gamma_calc = norm.pdf(d1,0,1)/(S*sigma*np.sqrt(T)) # whether it's a call or a put
    return gamma_calc, gamma(type, S, K, T, r, sigma)
  except:
    print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
```

```python
# 3.3 Vega calculus --- sensitivity of option price to changes in volatility.

def vega_calc(r,s,K,T,sigma, type="c"):
  "Calculate Vega of an option"
  d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  try:
    vega_calc = S*norm.pdf(d1,0,1)*np.sqrt(T) # whether it's a call or a put
    return vega_calc*0.01, vega(type, S, K, T, r, sigma)
  except:
    print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
```

```python
# 3.4 Theta calculus --- sensitivity of option price to the passage of time (time decay).

def theta_calc(r,s,K,T,sigma, type="c"):
  "Calculate Theta of an option"
  d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  try:
    if type == "c": # if it's a call
      theta_calc = -S*norm.pdf(d1,0,1)*sigma/(2*np.sqrt(T)) - r*K*np.exp(-r*T)*norm.cdf(d2,0,1)
    elif type =="p": # if it's a put
      theta_calc = -S*norm.pdf(d1,0,1)*sigma/(2*np.sqrt(T)) + r*K*np.exp(-r*T)*norm.cdf(-d2,0,1)
    return theta_calc*0.01, theta(type, S, K, T, r, sigma)
  except:
    print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
```

```python
# 3.5 Rho calculus --- sensitivity of option price to changes in interest rates.

def rho_calc(r,s,K,T,sigma, type="c"):
  "Calculate Rho of an option"
  d1 = (np.log(S/K) + (r+sigma**2/2)*T)/(sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  try:
    if type == "c": # if it's a call
      rho_calc = K*T*np.exp(-r*T)*norm.cdf(d2,0,1)
    elif type =="p": # if it's a put
      rho_calc = -K*T*np.exp(-r*T)*norm.cdf(-d2,0,1)
    return rho_calc*0.01, rho(type, S, K, T, r, sigma)
  except:
    print("Please confirm option type, 'c' if it's a Call or 'p' if it's a put")
```

```python
# 4. All combined
# 4.1 For the put

option_type="p"

print("Option price: ", [round(x,3) for x in blackscholes(r, S, K, T, sigma, option_type)])
print("Delta : ", [round(x,3) for x in delta_calc(r, S, K, T, sigma, option_type)])
print("Gamma :",[round(x,3) for x in gamma_calc(r, S, K, T, sigma, option_type)])
print("Vega :",[round(x,3) for x in vega_calc(r, S, K, T, sigma, option_type)])
print("Theta :",[round(x,3) for x in theta_calc(r, S, K, T, sigma, option_type)])
print("Rho :",[round(x,3) for x in rho_calc(r, S, K, T, sigma, option_type)])
```

```python
# 4.2 For the call

option_type="c"

print("Option price: ", [round(x,3) for x in blackscholes(r, S, K, T, sigma, option_type)])
print("Delta : ", [round(x,3) for x in delta_calc(r, S, K, T, sigma, option_type)])
print("Gamma :",[round(x,3) for x in gamma_calc(r, S, K, T, sigma, option_type)])
print("Vega :",[round(x,3) for x in vega_calc(r, S, K, T, sigma, option_type)])
print("Theta :",[round(x,3) for x in theta_calc(r, S, K, T, sigma, option_type)])
print("Rho :",[round(x,3) for x in rho_calc(r, S, K, T, sigma, option_type)])
```

**------Overview------**

Since the underlying price ($30) is significantly lower than the strike price ($40), the Put option is deeply In-The-Money (ITM), while the Call option is Out-of-The-Money (OTM). This explains the stark difference in option prices: $10.251 for the Put versus $0.513 for the Call.

**----> Interpretation of the Greeks :**

**Delta (Δ) —** Underlying Price Sensitivity
Delta measures the expected change in option price per $1.00 change in the underlying asset.

- Put (Δ=−0.849): Close to −1. A $1.00 increase in the underlying asset decreases the Put's value by approximately $0.85. The Put behaves similarly to a short position of roughly 0.85 shares, which is standard for a deeply ITM put.

- Call (Δ=0.151): Low and positive. A $1.00 increase in the underlying increases the Call's value by only $0.15. Being OTM, it has a low probability of finishing in-the-money.

- (Note: Put-Call Delta parity holds true here: 0.151−(−0.849)=1.00).

**Gamma (Γ) —** Delta Acceleration
Gamma measures the rate of change of Delta with respect to changes in the underlying asset. It is identical for both the Call and the Put.

- Value (Γ=0.032): A moderate Gamma indicates that Delta is relatively stable against small price fluctuations. Gamma peaks when the option is at-the-money (S=K). Here, since the underlying (30) is some distance away from the strike (40), Delta will only begin accelerating rapidly if the underlying approaches the 40 level.

**Vega (ν) —** Volatility Sensitivity
Vega measures the impact of a 1% (0.01) change in implied volatility (
σ). It is identical for both options.

- Value (ν=0.057): If volatility increases from 30% to 31%, both the Call and the Put prices will increase by approximately $0.057. Higher volatility increases the uncertainty and probability of large price swings, benefiting both option holders (especially the OTM Call, which relies heavily on volatility value).

**Theta (Θ) —** Time Decay
Theta measures the daily loss of option value due to the passage of time.

- Put (Θ≈−0.009 to −0.003) & Call (Θ≈−0.013 to −0.004):

- Values are negative, as expected for long option positions. Time decay works against the buyer because time value erodes as expiration approaches.

- The Call loses value slightly faster in relative terms due to its near-zero intrinsic value.

**Rho (ρ) —** Interest Rate Sensitivity
Rho measures the sensitivity of the option price to a 1% change in the risk-free interest rate.

- Put (ρ=−0.235): Negative. Higher interest rates decrease the present value of the strike price, which negatively impacts the Put.

- Call (ρ=0.026): Positive. Higher interest rates decrease the present value of the strike price, which benefits the Call.

**------Risk Profile Summary------**

In-The-Money Put (S=30,K=40): The price is predominantly intrinsic value (~$10 intrinsic value vs. ~$0.25 time value). The primary risk exposure is a sharp rebound in the underlying asset (high absolute Delta).

Out-of-The-Money Call (S=30,K=40): A speculative position consisting purely of time value. It is highly sensitive to underlying acceleration (Gamma) and volatility shifts (Vega), while constantly fighting time decay (Theta).
